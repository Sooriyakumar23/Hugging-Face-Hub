# 1. Import the Required Library

```python
from transformers import pipeline
```

# 2. Create a Text Generation Pipeline

```python
generator = pipeline(
    task = "text-generation",
    model = "Qwen/Qwen2.5-1.5B-Instruct",
    device_map = "auto"
)
```

# 3. Define Prompt

```python
prompt = "Explain AI."

prompt_list = [
    "Explain the classification task in simple language",
    "Explain the text generation task in simple language",
    "Explain text summarization task in simple language",
]
```

# 4. Do the Inference

### 4.1. Single Prompt

```python
result = generator(prompt, max_new_tokens=80, temperature=0.7)
```

```python
result
```

```python
result[0]['generated_text']
```

### 4.2. Multiple Prompts

```python
result = generator(prompt_list, max_new_tokens=80, temperature=0.7)
```
